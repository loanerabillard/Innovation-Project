```python
import os
import pandas as pd

def print_csv_fields_with_examples(directory):
    # Check if the directory exists
    if not os.path.isdir(directory):
        print(f"The directory {directory} does not exist.")
        return
    
    # Print the directory
    print(f"Directory: {directory}\n")

    # Iterate through each file in the directory
    for filename in os.listdir(directory):
        # Construct the full file path
        file_path = os.path.join(directory, filename)
        
        # Check if the file is a CSV file
        if filename.endswith(".csv") and os.path.isfile(file_path):
            try:
                # Read the CSV file using pandas
                df = pd.read_csv(file_path)
                
                # Get the columns of the dataframe
                columns = df.columns.tolist()
                
                # Print the name of the CSV file and its fields
                print(f"CSV File: {filename}")
                print("Fields:", columns)
                
                # Print one example row
                if not df.empty:
                    print("Example row:")
                    print(df.iloc[0].to_dict())
                else:
                    print("The file is empty.")
                
                print()
            except Exception as e:
                print(f"Could not read {filename} due to an error: {e}")
```

```python
import pandas as pd
import numpy as np

# Load data from CSV files
events_df = pd.read_csv('data/matches_csv/events.csv')
player_statistics_df = pd.read_csv('data/player_csv/player_statistics.csv')
pourcentage_victoire_df = pd.read_csv('data/player_csv/pourcentage_victoire.csv')

# Surface map (example)
surface_map = {
    'ATP ATP Cup': 'Hard',
    # Add other tournament mappings
}

# Function to calculate win probability
def calculate_win_probability(row, player_stats, surface_map):
    player1_key = row['first_player_key']
    player2_key = row['second_player_key']
    surface = surface_map.get(row['tournament_name'], 'Unknown')
    
    player1_stats = player_stats.get(player1_key)
    player2_stats = player_stats.get(player2_key)
    
    # If any player stats are missing, return None for probabilities
    if player1_stats is None or player2_stats is None:
        return None, None
    
    # Extract necessary statistics
    rank_A = player1_stats.get('rank', 0)
    rank_B = player2_stats.get('rank', 0)
    age_A = player1_stats.get('age', 27)
    win_percent_A = player1_stats.get('general_win_percent', 0)
    win_percent_B = player2_stats.get('general_win_percent', 0)
    surface_win_percent_A = player1_stats.get(f'{surface.lower()}_win_percent', 0)
    surface_win_percent_B = player2_stats.get(f'{surface.lower()}_win_percent', 0)
    
    # Previous encounter win percentage
    prev_encounter_percent_A = player1_stats.get(f'vs_{player2_key}_win_percent', 0)
    prev_encounter_percent_B = player2_stats.get(f'vs_{player1_key}_win_percent', 0)

    epsilon = np.random.uniform(0.001, 0.01)
    g_age_A = 1 / (abs(age_A - 27) + 1)
    
    # Check if players have met before
    if prev_encounter_percent_A > 0 or prev_encounter_percent_B > 0:
        Z_A = (0.1 * (rank_B - rank_A) + 
               0.4 * prev_encounter_percent_A + 
               0.3 * surface_win_percent_A + 
               0.2 * win_percent_A + 
               epsilon * g_age_A)
        Z_B = (0.1 * (rank_A - rank_B) + 
               0.4 * prev_encounter_percent_B + 
               0.3 * surface_win_percent_B + 
               0.2 * win_percent_B + 
               epsilon * (1 / (abs(player2_stats.get('age', 27) - 27) + 1)))
    else:
        Z_A = (0.1 * (rank_B - rank_A) + 
               0.3 * surface_win_percent_A + 
               0.2 * win_percent_A + 
               epsilon * g_age_A)
        Z_B = (0.1 * (rank_A - rank_B) + 
               0.3 * surface_win_percent_B + 
               0.2 * win_percent_B + 
               epsilon * (1 / (abs(player2_stats.get('age', 27) - 27) + 1)))
    
    return Z_A, Z_B

# Extract relevant player statistics into a dictionary
player_stats = {}
for _, row in player_statistics_df.iterrows():
    player_key = row['player_key']
    player_stats[player_key] = {
        'rank': row['rank'],
        'age': 2022 - int(row['player_bday'][-4:]),  # Assuming current year is 2022
        'general_win_percent': row['matches_won'] / (row['matches_won'] + row['matches_lost']) if (row['matches_won'] + row['matches_lost']) > 0 else 0,
        'hard_win_percent': row['hard_won'] / (row['hard_won'] + row['hard_lost']) if (row['hard_won'] + row['hard_lost']) > 0 else 0,
        'clay_win_percent': row['clay_won'] / (row['clay_won'] + row['clay_lost']) if (row['clay_won'] + row['clay_lost']) > 0 else 0,
        'grass_win_percent': row['grass_won'] / (row['grass_won'] + row['grass_lost']) if (row['grass_won'] + row['grass_lost']) > 0 else 0,
        f'vs_{row["player_key"]}_win_percent': 0  # Placeholder for previous encounter win percentage
    }

# Update previous encounter win percentages from pourcentage_victoire.csv
for _, row in pourcentage_victoire_df.iterrows():
    joueur_key = row['player_key']
    adversaire_key = player_statistics_df[player_statistics_df['player_name'] == row['Adversaire']]['player_key'].values[0]
    if joueur_key in player_stats:
        player_stats[joueur_key][f'vs_{adversaire_key}_win_percent'] = row['% Victoire']

# Calculate win probabilities for each event
win_probabilities = []
for _, row in events_df.iterrows():
    Z_A, Z_B = calculate_win_probability(row, player_stats, surface_map)
    if Z_A is None or Z_B is None:
        win_prob_player1 = None
        win_prob_player2 = None
    else:
        win_prob_player1 = 1 / (1 + np.exp(-(Z_A - Z_B)))
        win_prob_player2 = 1 / (1 + np.exp(-(Z_B - Z_A)))
    
    win_probabilities.append({
        'Joueur': row['event_first_player'],
        'Adversaire': row['event_second_player'],
        'Surface': surface_map.get(row['tournament_name'], 'Unknown'),
        'Win Probability Joueur 1': win_prob_player1,
        'Win Probability Joueur 2': win_prob_player2
    })

# Convert the results to a DataFrame for better visualization
win_probabilities_df = pd.DataFrame(win_probabilities)

win_probabilities_df
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, roc_auc_score
import matplotlib.pyplot as plt

# Check if GPU is available (optional)
def check_gpu():
    try:
        import torch
        if torch.cuda.is_available():
            print("GPU is available.")
        else:
            print("GPU is not available.")
    except ImportError:
        print("torch is not installed. Cannot check GPU availability.")

check_gpu()

# Load CSV files
events_df = pd.read_csv('data/matches_csv/events.csv')
points_df = pd.read_csv('data/matches_csv/points.csv')
scores_df = pd.read_csv('data/matches_csv/scores.csv')
sets_df = pd.read_csv('data/matches_csv/sets.csv')
player_statistics_df = pd.read_csv('data/player_csv/player_statistics.csv')

# Merge events with scores
events_scores_df = pd.merge(events_df, scores_df, on='event_key', how='left')

# Merge events with player statistics for both players
events_scores_df = pd.merge(events_scores_df, player_statistics_df, left_on='first_player_key', right_on='player_key', how='left', suffixes=('', '_first'))
events_scores_df = pd.merge(events_scores_df, player_statistics_df, left_on='second_player_key', right_on='player_key', how='left', suffixes=('', '_second'))

# Drop unnecessary columns
events_scores_df = events_scores_df.drop(columns=['player_key', 'player_name', 'player_logo'])

# Feature engineering
events_scores_df['rank_difference'] = events_scores_df['rank'] - events_scores_df['rank_second']
events_scores_df['titles_difference'] = events_scores_df['titles'] - events_scores_df['titles_second']
events_scores_df['matches_won_difference'] = events_scores_df['matches_won'] - events_scores_df['matches_won_second']
events_scores_df['matches_lost_difference'] = events_scores_df['matches_lost'] - events_scores_df['matches_lost_second']

# Label encoding
label_encoder = LabelEncoder()
events_scores_df['event_type_type'] = label_encoder.fit_transform(events_scores_df['event_type_type'])
events_scores_df['tournament_name'] = label_encoder.fit_transform(events_scores_df['tournament_name'])
events_scores_df['event_winner'] = events_scores_df['event_winner'].map({'First Player': 1, 'Second Player': 0})

# Select features and target
features = events_scores_df.drop(columns=['event_winner', 'event_final_result', 'event_game_result', 'event_status', 'event_date', 'event_time', 'event_first_player', 'event_second_player', 'event_first_player_logo', 'event_second_player_logo', 'tournament_round', 'event_live', 'tournament_key', 'tournament_season'])
target = events_scores_df['event_winner']

# Split the data
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)

# Initialize the model
model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)
y_pred_prob = model.predict_proba(X_test)[:, 1]

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred_prob)

print(f'Accuracy: {accuracy}')
print(f'ROC AUC: {roc_auc}')

# Feature importance
feature_importances = model.feature_importances_
features_list = features.columns

# Plot feature importance
plt.figure(figsize=(10, 8))
plt.barh(features_list, feature_importances)
plt.xlabel('Feature Importance')
plt.title('Feature Importance in RandomForest Model')
plt.show()

# Example prediction for new data
new_event_data = X_test.iloc[0].values.reshape(1, -1)
probability_of_victory = model.predict_proba(new_event_data)

print(f'Probability of victory for first player: {probability_of_victory[0][1]}')
print(f'Probability of victory for second player: {probability_of_victory[0][0]}')
```
